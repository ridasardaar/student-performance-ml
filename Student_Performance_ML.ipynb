# 🎓 Student Performance Prediction

## 📌 Problem Statement
Can we predict a student’s final grade (G3) using demographic, academic, and lifestyle features?
## 🎯 Objective
- Build and compare multiple ML models
- Identify key factors affecting student performance
- Evaluate model performance using R² score

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score 
```

## 📊 Dataset
- Source: UCI Machine Learning Repository
- Records: 395 students
- Features: 30+ variables (academic, social, personal)

```python
data = pd.read_csv("student-mat.csv",sep=";")
data.head() 
```

## Data Exploration
- Total rows and columns
- Data types
- Initial observations

```python
data.shape
data.info()
data.describe()
```

### Observations
- Both numerical and categorical features exist
- no missing values found


## ❓ Key Questions

- Which factors most influence student performance?
- Can lifestyle features predict academic success?
- Which model performs best on this dataset?

## Data Cleaning

### ⚠️ Data Leakage
G1(1st exam's grade) and G2(2nd exam's grade) were removed because they are previous exam scores and strongly correlated with G3(final grade).
Including them would result in unrealistically high model performance.

```python
data = data.drop(['G1', 'G2'], axis=1)
```

### 🔄 Feature Preparation & Encoding

- Separated input (X) and target (y)
- Categorical variables were converted into numerical format using one-hot encoding. 
    This allows machine learning models to process non-numeric data.


```python
X = data.drop('G3', axis=1)
y = data['G3']

X = pd.get_dummies(X)
### to convert boolean values to 0, 1
X = X.astype(int) 
```

After preprocessing, multiple regression models were trained and compared to evaluate their ability to predict student performance.

### Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

## Model Training

### Linear Regression

```python
lr = LinearRegression()
lr.fit(X_train, y_train)

y_pred_lr = lr.predict(X_test)
r2_lr = r2_score(y_test, y_pred_lr)
r2_lr
```

### 📈 Interpretation
Linear Regression performed poorly and achieved R² score of approximately 0.20.

This indicates weak linear relationship between features and target variable (G3). The model was able to capture only basic trends in the data and struggled with more complex relationships

### Decision Tree

```python
dt = DecisionTreeRegressor(max_depth=3)
dt.fit(X_train, y_train)

y_pred_dt = dt.predict(X_test)
r2_dt = r2_score(y_test, y_pred_dt)
r2_dt
```

### 🌳 Interpretation

Decision Tree achieved the best performance with an R² score of approximately 0.39.

This suggests that the dataset contains non-linear relationships that tree-based models can capture more effectively than linear models. The model was able to learn useful decision rules from the student-related features.

### Random Forest

```python
rf = RandomForestRegressor(random_state=42)
rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)
r2_rf = r2_score(y_test, y_pred_rf)
r2_rf
```

### 🌲 Interpretation

Random Forest achieved an R² score of approximately 0.26.

Although Random Forest is generally more robust than a single Decision Tree, its performance in this case was lower. This may indicate that the dataset is relatively small and that averaging multiple trees reduced some useful patterns learned by the Decision Tree model.

## 📊 Overall Model Comparison

Among the evaluated models, Decision Tree achieved the highest R² score (~0.39), indicating that non-linear relationships exist within the dataset.

Linear Regression showed limited performance due to weak linear correlations between features and target values.

Random Forest improved upon Linear Regression but did not outperform the Decision Tree model in this case, possibly due to the relatively small dataset size and feature complexity.

```python
results = pd.DataFrame({
    'Model': ['Linear Regression', 'Decision Tree', 'Random Forest'],
    'R2 Score': [r2_lr, r2_dt, r2_rf]
})

results
```

## 🔍 Feature Importance Insights

Top features:
- absences → attendance impacts performance
- failures → strong negative academic signal
- goout → social activity affects study time
Surprisingly:
- studytime was not a top predictor, suggesting complexity in learning behavior

```python
feature_importance = pd.Series(rf.feature_importances_, index=X.columns)
feature_importance.sort_values(ascending=False).head(10).plot(kind='bar')
plt.title("Top Features")
plt.show()
```
